```python
import numpy as np
from scipy.constants import hbar, speed_of_light, Boltzmann, g
import math
```

Dipole Trap Depth

$U_{dipole} = \frac{\hbar \Gamma^2 I}{8\delta I_{sat}}$

```python
#Constants 

w_0 = 780 * 1e-9 #transition wavelength 
w = 1064*1e-09 #laser frequecy 
Rb_transition_freq  = 2*math.pi * 6.065 *1e6 #transition frequency Gamma
laser_power = 1 #Watt
laser_waist = 10e-6
SPEED_OF_LIGHT = 300000000
I_sat = 16.7 #W/m^2
mass_Rb = 1.443*1e-25 # kg mass of Rb 87
```

```python
#Computed Values

detuning = 2*math.pi*speed_of_light*(1/w_0 -1/w)
I_over_I_sat = 2*laser_power/(math.pi * laser_waist**2)/I_sat

U_dipole = hbar*Rb_transition_freq**2*I_over_I_sat/(8*detuning)
C = U_dipole * laser_waist**2

print(U_dipole, 'J')

print(U_dipole/Boltzmann*1000, 'mK')
```

Now find the ratio of the size of the horizontal and longitudinal extent of the dipole trap. This is just isoenergy line?

```python
dz_dr = np.sqrt(2) * math.pi * laser_waist / w

print(dz_dr, 'times larger in longitudinal coordinate than transverse state.')
```

Now find the maximum beam waist where the dipole trap can support Rb atoms against gravity, at the max force of $r=w_0/2$

```python
gravity_force = mass_Rb * g

max_waist = (2 * C * np.exp(-1/2) / gravity_force)**(1/3)

print(max_waist * 1000000, 'um')
```

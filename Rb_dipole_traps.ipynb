$U_{dipole} = \frac{\hbar \Gamma^2 I}{8\delta I_{sat}}$

```python
%%render params
w_0 = 780 * 1e-9 *m#transition wavelength 
w = 852*1e-09 *m #laser wavelength
Rb_transition_freq  = 2*pi * 6.065 *1e6 *Hz #transition frequency Gamma
laser_power = 1 *W#Watt
laser_waist =55e-6 *m #laser waist
c = 300000000 *m/s
I_sat = 16.7 *W/m**2 #saturation intensity
mass_Rb = 1.443*1e-25 *kg # kg mass of Rb 87
hbar = 1.054571817e-34 * (kg * m**2 / s)
k_B = 1.380649e-23 * si.kg * si.m**2 / (si.s**2 * si.K)
g=9.8*m/s**2
```

```python
detuning = 2*pi*c*(1/w_0 - 1/w)
```

```python
%%render
I_over_I_sat = 2*laser_power/(pi * laser_waist**2)/I_sat #ratio of $\frac{I}{I_{sat}}$
```

```python
U_dipole = hbar*Rb_transition_freq**2*I_over_I_sat/(8*detuning)
```
